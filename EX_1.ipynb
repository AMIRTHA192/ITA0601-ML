```python
import pandas as pd

data = pd.read_csv('/content/enjoysport.csv')

print("Training Data:")
print(data)

hypothesis = None

for _, row in data.iterrows():
    if str(row['enjoysport']).strip().lower() == 'yes':
        attributes = row.drop('enjoysport').tolist()

        if hypothesis is None:
            hypothesis = attributes.copy()
        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != attributes[i]:
                    hypothesis[i] = '?'

print("\nMost Specific Hypothesis:")
print(hypothesis)
```

```python
import pandas as pd

data = pd.read_csv('/content/enjoysport.csv')

X = data.iloc[:, :-1].values.tolist()
y = data.iloc[:, -1].astype(str).str.strip().str.lower().tolist()

n = len(X[0])
S = ['0'] * n
G = [['?'] * n]

for x, target in zip(X, y):
    if target == 'yes':
        G = [g for g in G if all(
            g[i] == '?' or g[i] == x[i] for i in range(n)
        )]

        for i in range(n):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'
    else:
        new_G = []
        for g in G:
            if all(g[i] == '?' or g[i] == x[i] for i in range(n)):
                for i in range(n):
                    if S[i] != '?' and S[i] != '0' and S[i] != x[i]:
                        h = g.copy()
                        h[i] = S[i]
                        new_G.append(h)
            else:
                new_G.append(g)

        G = new_G

print("Final Specific Hypothesis (S):")
print(S)

print("\nFinal General Hypothesis (G):")
for g in G:
    print(g)
```
